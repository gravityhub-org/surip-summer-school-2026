# Introduction to Machine Learning with PyTorch

This repository is for a one-hour interactive workshop on training a small image classifier with PyTorch.

The workshop follows this idea:

```text
Notebook = understand and inspect
Script   = reusable ML project code
Terminal = train the model
Config   = run experiments
Notebook = evaluate and interpret the result
```

The dataset is a small CIFAR-10 subset exported into local image folders.

## Expected repository structure

```text
intro-ml-pytorch-workshop-cifar/
├── README.ipynb
├── SETUP.ipynb
├── requirements.txt
├── check_setup.py
├── config.yaml
├── prepare_cifar10_subset.py
├── data.py
├── model.py
├── train.py
├── utils.py
├── notebooks/
│   ├── 01_dataset_exploration.ipynb
│   ├── 02_model_architecture.ipynb
│   └── 03_evaluate_run.ipynb
└── outputs/
    └── runs/
```

The dataset folder `data_cifar/` will be created after running `prepare_cifar10_subset.py`.

## Student workflow

1. Follow `SETUP.ipynb` to create the Python environment.
2. Run the CIFAR preparation script:

```bash
python prepare_cifar10_subset.py --train-per-class 400 --val-per-class 100
```

3. Check that the setup works:

```bash
python check_setup.py
```

4. Open `notebooks/01_dataset_exploration.ipynb`.
   - Inspect image labels.
   - Print image batch shape `[B, C, H, W]`.
   - Understand the working DataLoader code.

5. Complete the TODOs in `data.py`.

6. Open `notebooks/02_model_architecture.ipynb`.
   - Inspect the CNN architecture.
   - Visualize the tensor-shape flow through the model.
   - Inspect logits, softmax probabilities, and predicted classes.

7. Run a dry run:

```bash
python train.py --config config.yaml --dry-run
```

8. Train the model:

```bash
python train.py --config config.yaml
```

9. Modify `config.yaml`, rerun training, and compare results.

10. Open `notebooks/03_evaluate_run.ipynb` to load the latest run and show the saved plots.

## What students submit

Each student/group should submit one training run containing:

```text
outputs/runs/<run_name>/config_used.yaml
outputs/runs/<run_name>/logs/metrics.csv
outputs/runs/<run_name>/figures/loss_accuracy_curve.png
outputs/runs/<run_name>/figures/confusion_matrix.png
outputs/runs/<run_name>/figures/prediction_grid.png
```

They should also write 2–3 sentences describing what they changed and what happened.

Example:

> I changed the learning rate from 0.001 to 0.01. The training loss decreased faster at first, but the validation accuracy became less stable. This suggests that the larger learning rate may be too aggressive.
