# Setup instructions

The workshop is designed to run on CPU. A GPU is optional.

Use Terminal on macOS/Linux or PowerShell on Windows. Do not run the installation commands inside a notebook cell.

## 1. Create a virtual environment

### macOS / Linux

```bash
cd intro-ml-pytorch-workshop-cifar
python -m venv .venv-ml
source .venv-ml/bin/activate
python -m pip install --upgrade pip
```

### Windows PowerShell

```powershell
cd intro-ml-pytorch-workshop-cifar
python -m venv .venv-ml
.venv-ml\Scripts\Activate.ps1
python -m pip install --upgrade pip
```

If PowerShell blocks activation, run:

```powershell
Set-ExecutionPolicy -Scope CurrentUser RemoteSigned
```

Then reopen PowerShell and activate again.

## 2. Install PyTorch

Install the CPU version by default. The workshop does not require CUDA.

Go to the official PyTorch install selector and choose:

```text
Stable
Your operating system
Pip
Python
CPU
```

Then copy the generated command into Terminal/PowerShell.

For many systems, this may look like:

```bash
python -m pip install torch torchvision
```

If you already know you have an NVIDIA GPU and want CUDA, use the PyTorch selector and choose the appropriate CUDA option. This is optional.

## 3. Install the remaining packages

```bash
python -m pip install -r requirements.txt
```

Register the notebook kernel:

```bash
python -m ipykernel install --user --name intro-ml-workshop --display-name "ML Workshop"
```

## 4. Prepare the CIFAR subset

```bash
python prepare_cifar10_subset.py --train-per-class 400 --val-per-class 100
```

For a very quick test, use the smaller version:

```bash
python prepare_cifar10_subset.py --train-per-class 100 --val-per-class 25 --overwrite
```

This creates:

```text
data_cifar/
├── train/
│   ├── airplane/
│   ├── automobile/
│   ├── cat/
│   └── ship/
└── val/
    ├── airplane/
    ├── automobile/
    ├── cat/
    └── ship/
```

## 5. Check the setup

```bash
python check_setup.py
```

Expected output includes:

```text
PyTorch version: ...
Detected device: cpu / cuda / mps
Dataset classes: ['airplane', 'automobile', 'cat', 'ship']
One image shape: (3, 32, 32)
Setup check passed.
```

If it says `Detected device: cpu`, that is correct. The workshop is CPU-compatible.
